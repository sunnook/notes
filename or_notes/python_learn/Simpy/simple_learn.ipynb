```python

def generator(n):
    while True:
        for j in range(n):
            yield j

def function(n):
    for j in range(n):
        return j

ourGen = generator(3)
ourFun = function(3)


print(ourGen.__next__())
print(ourGen.__next__())
print(next(ourGen))
print(next(ourGen))
print(next(ourGen))
print(next(ourGen))
print('    ')
print(ourFun)
print(ourFun)
```

```python
# 红绿灯案例：

import simpy


def main():
    env = simpy.Environment()
    env.process(traffic_light(env))
    env.run(until=120)  # 设置最多运行的时间
    print("Simulation complete")

# define a process
def traffic_light(env):
    while True:
        print("Light turned GREEN at t="+str(env.now))
        yield env.timeout(30)  # 并不会真的等待30s，而是假装等待了30s的“仿真时间”，实际上是立跳到下面
        print("Light turned YELOOW at t="+str(env.now))
        yield env.timeout(5)
        print("Light turned RED at t="+str(env.now))
        yield env.timeout(20)

if __name__ == '__main__':
    main()






```
