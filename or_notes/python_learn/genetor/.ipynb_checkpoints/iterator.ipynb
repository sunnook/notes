一、迭代器

迭代器是包含__iter__和__next__（魔法方法）的对象
    可迭代对象的iter返回一个迭代器；迭代器的iter返回本身
    next返回下一个迭代器；直到raise StopIteration

python的for循环：
1.调用iterables的__iter__获取iterator
2.调用iterator的__next__函数，获取下一个值
    能获取到就返回下一个值
    否则，退出for循环

```python
mystr = '123'
for s in mystr:
    print(s)

mylst = [1,2,3]
for i in mylst:
    print(i)

myint = 123
print(hasattr(mystr, '__iter__'))  # 有__iter__方法的对象，就是iterables（可迭代对象）
print(hasattr(mylst, '__iter__'))
print(hasattr(myint, '__iter__'))

#for n in myint:  # 报错
#    print(n)
```

```python
# 上面的for循环等价于next(it)：
it = iter(mystr) # 等价于 mystr.__iter__()
while True:
    try:
        print(next(it))
    except StopIteration:
        break
```

```python
# 使用f.readlines()一次读取文件所有行的数据

def process_line(line):  # 处理每一行
    pass

filepath = './logfile.log'

import tracemalloc   # 内存查看
tracemalloc.start()  # 开始跟踪内存分配

with open(filepath,'r') as f:
    lines = f.readlines()    # f.readlines(), 这里已经全部加载内存中了

current, peak = tracemalloc.get_traced_memory()  # 内存，当前内存和峰值内存使用
print(f"Current memory usage: {current/1024**2} MB")
print(f"Peak memory usage: {current/1024**2} MB")

for line in lines:
    process_line(line)


current, peak = tracemalloc.get_traced_memory()  # 内存，当前内存和峰值内存使用
print(f"Current memory usage: {current/1024**2} MB")
print(f"Peak memory usage: {current/1024**2} MB")
tracemalloc.stop()  # 停止跟踪内存分配
```

```python
# 定义一个迭代器，内部定义__next__函数，一行行取出log中文件
#     __next__函数调用readline()，读取一行

class LineIterator:
    def __init__(self, filepath):
        self.file = open(filepath, 'r')
    
    def __iter__(self):  # 有iter和next方法的类叫做迭代器类
        return self
    
    def __next__(self):
        line = self.file.readline()  # file.readline(), 这里是一行行加载
        # if line:  # 仅读取
        #     return line
        # else:
        #     self.file.close()
        #     raise StopIteration
        while line:
            if line.split('|')[2].strip() == 'create21':
                return line
            line = self.file.readline()
        self.file.close()
        raise StopIteration

import tracemalloc   # 内存查看
tracemalloc.start()  # 开始跟踪内存分配

line_iter = LineIterator(filepath)
for line in line_iter:   # for循环会自动调用__iter__和__next__
    print(line)

current, peak = tracemalloc.get_traced_memory()  # 内存，当前内存和峰值内存使用
print(f"Current memory usage: {current/1024**2} MB")
print(f"Peak memory usage: {current/1024**2} MB")
tracemalloc.stop()  # 停止跟踪内存分配
```

二、生成器

可以理解为迭代器的简单实现；
有2种写法：
    1.生成器函数 
    2.生成器表达式

在yeild处产生一个值，然后退出函数，下次进来时又从yield处继续；

```python
def generator(n):
    for i in range(n):
        print('before yield')
        yield i    # 通过yield关键字，生成器会自动产生 next 和iter方法
        print('after yield')

gen = generator(3)

print(next(gen))
print('---')

for i in gen:
    print(i)
```

```python
# 生成器函数
def line_generator(filepath):
    with open(filepath, 'r') as file:
        for line in file:
            if line.split('|')[2].strip()=='create21':
                yield line
            else:
                continue

line_gen = line_generator(filepath)
for line in line_gen:
    print(line)
```

迭代器，和生成器的好处：
1.惰性计算，只有在处理这个元素时，才去处理他-->节省内存资源
2.生成器没有终点

```python
# 生成器函数例子：斐波那契数列
def fib_generate():
    cur,nxt = 0,1
    while True:
        yield cur
        cur, nxt = nxt, cur+nxt

fib_gen = fib_generate()
for _ in range(10):
    print(next(fib_gen))
```

```python
# 生成器函数
def multi_generator(x):
    num = 1
    while True:
        res = num*x
        yield res
        num = num+1

multi_gen = multi_generator(2)
print(next(multi_gen))
print(next(multi_gen))
print(next(multi_gen))
print(next(multi_gen))
```


生成器函数和生成器表达式都是创建Python生成器的方式，但它们有一些关键的不同之处：

生成器函数（Generator Function）
定义: 使用 yield 语句来返回一系列值的函数被称为生成器函数。
特点:
    在函数体中可以有一个或多个 yield 语句。
    调用生成器函数不会立即执行函数体内的代码；而是返回一个可迭代对象（即生成器对象）。
    当生成器的 __next__() 方法被调用时，函数内部的代码从上次停止的位置继续执行，直到找到下一个 yield 语句，然后返回一个值，并再次暂停。
    生成器函数可以使用 send() 方法向 yield 表达式传递值。
    生成器函数可以拥有复杂的逻辑，并且可以有条件地生成值。

生成器表达式（Generator Expression）
定义: 类似于列表推导式（list comprehension），但使用圆括号而不是方括号。
特点:
    更加简洁和紧凑，适合简单的操作。
    生成器表达式是在一个表达式的上下文中定义的，通常用于需要一个可迭代对象的地方。
    生成器表达式不会占用太多内存，因为它是在需要的时候才生成下一个项。
    由于生成器表达式是表达式的一部分，所以不能包含条件分支、循环控制结构等复杂逻辑。
    生成器表达式主要用于生成序列，而不是执行复杂的逻辑。

示例对比
下面是一些生成器函数和生成器表达式的示例，以及如何使用它们：

```python
# 生成器函数

def square_numbers(nums):  # 生成器函数
    for n in nums:
        yield n * n

numbers = [1, 2, 3, 4]
squares = square_numbers(numbers)
for square in squares:
    print(square)
```

```python
# 生成器表达式

numbers = [1, 2, 3, 4]
squares = (n * n for n in numbers)  # 生成器表达式
for square in squares:
    print(square)
```
